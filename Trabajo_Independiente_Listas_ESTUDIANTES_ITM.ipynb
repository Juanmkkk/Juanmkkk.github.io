### 3.1 — Clasificación de calidad del aire
Clasifica cada medición de PM2.5 de una lista en 'Buena' (≤12), 'Moderada' (13-35) o 'Dañina' (>35), y cuenta cuántas hay de cada categoría.

```python
mediciones = [8, 15, 40, 22, 55]
buena = 0
moderada= 0
dañina = 0

for m in mediciones:
  if m <= 12:
    buena += 1

  elif m <=35 and m >= 15:
    moderada += 1

  else:
    dañina += 1

print(f"Buenas: {buena}, Moderadas: {moderada}, Dañinas: {dañina}")




```
